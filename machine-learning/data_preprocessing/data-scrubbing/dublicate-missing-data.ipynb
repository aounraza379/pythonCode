1.1 DATA SCRUBBING

```python
import numpy as np
import pandas as pd
```

1.1.1 Removing Duplicate Records

```python
data = {
    'ID': [1, 2, 3, 4, 1, 2],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Alice', 'Bob'],
    'Age': [25, 30, 35, 40, 25, 30],
    'Salary': [50000, 60000, 70000, 80000, 50000, 65000]
}
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:", df)

# Remove duplicate rows
df_no_dublicates = df.drop_duplicates()
print("DataFrame after removing duplicates:", df_no_dublicates)

# Remove duplicates based on specific columns
df_specific = df.drop_duplicates(subset=['ID', 'Name'])
print("DataFrame after removing duplicates based on ID and Name:", df_specific)
```

1.1.2 Handling Missing Data

```python
# data with missing values
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': [1, 2, 3, 4, 5],
    'Category': ['X', 'Y', np.nan, 'X', 'Y']
}
df = pd.DataFrame(data)

print("Data with missing values:")
print(df)
print("\nMissing values count:")
print(df.isnull().sum())

# Method 1: Deletion
df_dropna = df.dropna()  # Remove rows with any NaN
df_dropna_thresh = df.dropna(thresh=3)  # Keep rows with at least 3 non-NaN

# Method 2: Mean/Median imputation
df['A_imputed_mean'] = df['A'].fillna(df['A'].mean())
df['A_imputed_median'] = df['A'].fillna(df['A'].median())

# Method 3: Forward/Backward fill
df['B_ffill'] = df['B'].fillna(method='ffill')
df['B_bfill'] = df['B'].fillna(method='bfill')

# Method 4: Interpolation
df['A_interpolated'] = df['A'].interpolate(method='linear')

# Method 5: Categorical imputation
df['Category_imputed'] = df['Category'].fillna(df['Category'].mode()[0])

print("\nAfter various imputation methods:")
print(df)
```
